```python
import matplotlib.pyplot as plt
import numpy as np

# --- ১. LCG প্যারামিটার সেটআপ ---
m = 2**31 - 1  # Modulus (খুবই বড় প্রাইম নাম্বার, যাতে সংখ্যা সহজে রিপিট না হয়)
a = 1103515245  # Multiplier (ANSI C স্ট্যান্ডার্ড মান)
c = 12345  # Increment
seed = 42  # Initial Seed (X_0)

num_samples = 20  # আমরা ২০টি সংখ্যা তৈরি করে দেখব
lcg_numbers = []
normalized_numbers = []

# --- ২. LCG অ্যালগরিদম রান করা ---
current_X = seed
for _ in range(num_samples):
    current_X = (a * current_X + c) % m
    lcg_numbers.append(current_X)
    normalized_numbers.append(current_X / m)  # ০ থেকে ১ এর মধ্যে মান

# --- ৩. টেক্সট রিপোর্ট/টেবিল প্রিন্ট ---
print("=" * 50)
print(f"   LCG RANDOM NUMBER REPORT (Seed: {seed})")
print("=" * 50)
print(f"{'Index':<6} | {'Raw LCG Value (X_n)':<20} | {'Normalized (0 to 1)':<18}")
print("-" * 50)
for i in range(num_samples):
    print(f"{i+1:<6} | {lcg_numbers[i]:<20} | {normalized_numbers[i]:<18.4f}")
print("=" * 50)

# --- ৪. গ্রাফ প্রদর্শন (Scatter Plot) ---
plt.figure(figsize=(10, 5))
plt.scatter(range(1, num_samples + 1), normalized_numbers, color="#3498db", edgecolor="black", s=100, zorder=3)
plt.plot(range(1, num_samples + 1), normalized_numbers, color="#95a5a6", linestyle="--", alpha=0.5)
plt.title("Graph: Pseudo-Random Numbers Generated via LCG", fontsize=13, fontweight="bold", color="#2c3e50")
plt.xlabel("Sample Index", fontsize=11, fontweight="bold")
plt.ylabel("Generated Value (0.0 to 1.0)", fontsize=11, fontweight="bold")
plt.xlim(0, num_samples + 1)
plt.ylim(-0.05, 1.05)
plt.grid(True, linestyle="--", alpha=0.6)
plt.show()
```
