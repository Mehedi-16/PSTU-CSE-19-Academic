```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import scipy.stats as stats  # হ্যান্ড-ক্যালকুলেশনের বদলে স্ট্যান্ডার্ড লাইব্রেরি

# -----------------------------
# Part 1: λ = 5 calls/hour
# -----------------------------
lam = 5
x_values = np.arange(0, 11)
# ১ লাইনে নিখুঁত প্রবাবিলিটি ক্যালকুলেশন
probabilities = stats.poisson.pmf(x_values, mu=lam)

# Table Presentation
df = pd.DataFrame({
    "Number of Calls (x)": x_values,
    "P(X=x)": probabilities
})
print("=========================================")
print("       Poisson Distribution (λ = 5)")
print("=========================================")
print(df.round(6).to_string(index=False)) # index=False দিলে টেবিলটি দেখতে সুন্দর লাগে
print("=========================================\n")

# PMF Graph for λ = 5
plt.figure(figsize=(7, 4.5))
# কাস্টম কালার ও গ্রিড ডিজাইন
plt.bar(x_values, probabilities, color='#3498db', edgecolor='black', alpha=0.8, width=0.6)
plt.xlabel("Exact Number of Calls (x)", fontweight='bold')
plt.ylabel("Probability P(X=x)", fontweight='bold')
plt.title("Poisson PMF (λ = 5)", fontsize=11, fontweight='bold', color='#2c3e50')
plt.xticks(x_values)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

# -----------------------------
# Part 2: Compare λ=10 and λ=15
# -----------------------------
x = np.arange(0, 31)
pmf_10 = stats.poisson.pmf(x, mu=10)
pmf_15 = stats.poisson.pmf(x, mu=15)

plt.figure(figsize=(9, 4.5))
# লাইনগুলোকে একটু মোটা এবং কালারফুল করা হয়েছে
plt.plot(x, pmf_10, marker='o', linewidth=2, color='#e74c3c', label='λ = 10 (Simulated)')
plt.plot(x, pmf_15, marker='s', linewidth=2, color='#2ecc71', label='λ = 15 (Simulated)')

plt.xlabel("Number of Calls (x)", fontweight='bold')
plt.ylabel("Probability Density", fontweight='bold')
plt.title("Poisson PMF Curve Comparison", fontsize=11, fontweight='bold', color='#2c3e50')
plt.xticks(np.arange(0, 31, 2))
plt.legend(frameon=True, facecolor='white', edgecolor='#bdc3c7')
plt.grid(True, linestyle=':', alpha=0.6)
plt.tight_layout()
plt.show()
```
