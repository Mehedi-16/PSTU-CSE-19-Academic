```python
import numpy as np
import pandas as pd
import scipy.stats as stats
import matplotlib.pyplot as plt

# =========================================================================
# Task 1: Define Parameters and Outcomes (X values)
# =========================================================================
# p = Success probability (Chip passes the test)
# q = Failure probability (Chip is defective)
p = 0.85     
q = 1 - p    

# X values and their representations
x_outcomes = [1, 0]
representations = ["Success (Works Perfectly)", "Failure (Defective/Defect)"]

print("=" * 65)
print("       LAB 04: BERNOULLI DISTRIBUTION (QUALITY-CONTROL TEST)")
print("=" * 65)
print("Values of X and what they represent:")
print(f"X = 1 : {representations[0]} | Probability (p) = {p}")
print(f"X = 0 : {representations[1]} | Probability (q) = {q:.2f}")
print("-" * 65)

# =========================================================================
# Task 2: Calculating Probability of a Defect, P(X=0)
# =========================================================================
prob_defect = stats.bernoulli.pmf(0, p)
print(f"Probability of a defect, P(X=0)  : {prob_defect:.2f} ({prob_defect*100:.1f}%)")
print(f"Probability of a success, P(X=1) : {p:.2f} ({p*100:.1f}%)")
print("-" * 65)

# =========================================================================
# Task 3: Expected Metrics (Mean & Variance)
# =========================================================================
# সূত্র: E[X] = p
# সূত্র: Var(X) = p * q
expected_value_mean = p
variance_spread = p * q

print("--- EXPECTED METRICS ---")
print(f"Expected Value / Mean, E[X]      : {expected_value_mean:.2f}")
print(f"Variance / Spread, Var(X)        : {variance_spread:.4f}")
print("=" * 65)

# =========================================================================
# Visualizing the Bernoulli Distribution (PMF)
# =========================================================================
plt.figure(figsize=(8, 5))

# বার চার্ট তৈরি (০ এবং ১ এর জন্য)
x_axis = [0, 1]
pmf_values = [prob_defect, p]
colors = ['#e74c3c', '#2ecc71'] # লাল = ফেইলর, সবুজ = সাকসেস

bars = plt.bar(x_axis, pmf_values, color=colors, edgecolor='black', width=0.4, alpha=0.85)

# বারের ওপরে পার্সেন্টেজ লেবেল বসানো
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2.0, yval + 0.02, f"{yval*100:.1f}%", 
             ha='center', va='bottom', fontweight='bold', fontsize=11)

# গ্রাফ সুন্দর করার প্রফেশনাল সেটআপ
plt.title("Lab 04: Bernoulli Distribution for Microchip Quality Control", fontsize=12, fontweight='bold', color='#2c3e50')
plt.xlabel("Test Outcome (X)", fontsize=10, fontweight='bold')
plt.ylabel("Probability", fontsize=10, fontweight='bold')
plt.xticks(x_axis, ['Failure (X=0)', 'Success (X=1)'])
plt.ylim(0, 1.0)
plt.grid(axis='y', linestyle=':', alpha=0.6)

# তথ্য সংবলিত একটি টেক্সট বক্স
plt.text(0.2, 0.5, f"Expected Mean E[X] = {expected_value_mean:.2f}\nVariance Var(X) = {variance_spread:.4f}", 
         fontsize=10, fontweight='bold', bbox=dict(facecolor='white', alpha=0.8, boxstyle='round,pad=0.5'))

plt.tight_layout()
plt.show()
```
