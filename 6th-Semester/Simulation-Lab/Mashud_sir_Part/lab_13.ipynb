```python
import numpy as np
import matplotlib.pyplot as plt

# --- ১. প্যারামিটার এবং কনফিগুরেশন সেটআপ ---
NUM_SAMPLES = 5000  # ল্যাব ১৩ এর শর্তানুযায়ী ৫,০০০টি স্যাম্পল
LAMBDA = 0.5        # রেট প্যারামিটার (Rate Parameter)
np.random.seed(123) # ফিক্সড আউটপুট পাওয়ার জন্য সীড ভ্যালু

# --- ২. ইনভার্স ট্রান্সফর্ম মেথড প্রয়োগ ---
# ধাপ ক: ০ থেকে ১ এর মধ্যে সমানভাবে বিস্তৃত (Uniform) ৫,০০০টি র্যান্ডম সংখ্যা তৈরি
U = np.random.uniform(0, 1, NUM_SAMPLES)

# ধাপ খ: ইনভার্স ট্রান্সফর্ম সূত্র ব্যবহার করে এক্সপোনেনশিয়াল ভ্যারিয়েটে রূপান্তর
X = - (1 / LAMBDA) * np.log(U)

# --- ৩. টেক্সট রিপোর্ট ও পরিসংখ্যান প্রিন্ট ---
print("=" * 65)
print(f"      INVERSE TRANSFORM REPORT ({NUM_SAMPLES} Exponential Samples)")
print("=" * 65)
print(f"Mean of Generated Variates : {np.mean(X):.4f} (Theoretical: {1/LAMBDA:.1f})")
print(f"Variance of Variates       : {np.var(X):.4f} (Theoretical: {1/(LAMBDA**2):.1f})")
print("-" * 65)
print("First 5 Generated Values Samples:")
for i in range(5):
    print(f"Sample {i+1}: U = {U[i]:.4f} ---> X = {X[i]:.4f}")
print("=" * 65)

# --- ৪. গ্রাফ প্লট করা (Histogram + Theoretical PDF) ---
plt.figure(figsize=(10, 6))

# ৫,০০০ ডাটার হিস্টোগ্রাম (Probability Density বা ঘনত্বের আকারে)
plt.hist(X, bins=50, density=True, color="#9b59b6", edgecolor="white", alpha=0.75, label="Generated Data (5,000 samples)")

# আদর্শ তাত্ত্বিক লাইন তৈরি (Theoretical Exponential PDF Curve)
x_line = np.linspace(0, np.max(X), 1000)
pdf_line = LAMBDA * np.exp(-LAMBDA * x_line)
plt.plot(x_line, pdf_line, color="#e74c3c", linewidth=3, label="Theoretical PDF Curve")

# গ্রাফের লেবেল এবং টাইটেল সেট করা
plt.title(f"Lab 13: Histogram of {NUM_SAMPLES} Variates vs Theoretical PDF", fontsize=13, fontweight="bold", color="#2c3e50")
plt.xlabel("Value of Random Variate (X)", fontsize=11, fontweight="bold")
plt.ylabel("Probability Density", fontsize=11, fontweight="bold")
plt.legend(fontsize=11, frameon=True, shadow=True)
plt.grid(True, linestyle="--", alpha=0.5)

# লেআউট সুন্দরভাবে স্ক্রিনে ফিট করার জন্য
plt.tight_layout()
plt.show()
```
