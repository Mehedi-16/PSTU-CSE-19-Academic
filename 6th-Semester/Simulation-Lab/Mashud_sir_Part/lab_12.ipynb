```python
import numpy as np
import matplotlib.pyplot as plt

# --- ১. প্যারামিটার সেটআপ ---
num_samples = 100
lam = 0.5  # Rate parameter (lambda)
np.random.seed(42)  # ফিক্সড আউটপুটের জন্য সীড

# --- ২. ইনভার্স ট্রান্সফর্ম অ্যালগরিদম ---
# ধাপ ক: ০ থেকে ১ এর মধ্যে Uniform র্যান্ডম সংখ্যা তৈরি
U = np.random.uniform(0, 1, num_samples)

# ধাপ খ: ইনভার্স সূত্র প্রয়োগ (X = -1/lam * ln(U))
X = - (1 / lam) * np.log(U)

# --- ৩. টেক্সট রিপোর্ট প্রিন্ট ---
print("=" * 55)
print("   INVERSE TRANSFORM METHOD REPORT (Exponential)")
print("=" * 55)
print(f"{'Sample':<8} | {'Uniform U (0 to 1)':<20} | {'Generated Variate X':<18}")
print("-" * 55)
for i in range(10):  # প্রথম ১০টি স্যাম্পল প্রিন্ট করে দেখাচ্ছি
    print(f"{i+1:<8} | {U[i]:<20.4f} | {X[i]:<18.4f}")
print("... বাকি ৯০টি মান একইভাবে জেনারেট হয়েছে ...")
print("=" * 55)

# --- ৪. গ্রাফ প্রদর্শন (Probability Histogram) ---
plt.figure(figsize=(10, 5))
plt.hist(X, bins=15, density=True, color="#2ecc71", edgecolor="black", alpha=0.7, label="Generated Variates")

# আসল তাত্ত্বিক লাইন (Theoretical Probability Density Function)
x_axis = np.linspace(0, max(X), 100)
pdf = lam * np.exp(-lam * x_axis)
plt.plot(x_axis, pdf, color="#e74c3c", linewidth=2.5, label="Theoretical PDF (Exponential)")

plt.title("Graph: Random Variates via Inverse Transform Function", fontsize=13, fontweight="bold")
plt.xlabel("Value of Variate (X)", fontsize=11, fontweight="bold")
plt.ylabel("Probability Density", fontsize=11, fontweight="bold")
plt.legend()
plt.grid(axis='y', linestyle="--", alpha=0.5)
plt.show()
```
